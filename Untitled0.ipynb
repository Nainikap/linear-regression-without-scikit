```python
import numpy as np
import pandas as pd
import math
import operator
import matplotlib.pyplot as plt
%matplotlib inline
import io

from google.colab import files
uploaded = files.upload()

data = pd.read_csv(io.BytesIO(uploaded['test.csv']))
print(data.head())
```

```python
#algo for regression equation is:
# m = sigma(x-xbar)(y-ybar)/sigma(x-xbar)^2
#cost funciton => mean squared error
# cf = sigma(yhat-ybar)^2/sigma(y-ybar)^2

X = data['x'].values
Y = data['y'].values
xbar = np.mean(X)
ybar = np.mean(Y)

m = len(X)
numer=0
denom=0
for i in range(m):
  numer += (X[i]-xbar)*(Y[i]-ybar)
  denom += (X[i]-xbar)**2
m = numer/denom
c = ybar/(m*xbar)

print(m)
print(c)
```

```python
minx = np.min(X)-100
maxx = np.max(X)+100
x = np.linspace(minx, maxx,100)
y=c+m*x
plt.plot(x,y, color='#58b970', label='Regression Line')
plt.legend()
plt.scatter(X, Y, c='#ef5423', label='data points')
plt.show()
```

```python
sst = 0
ssr=0
val_count = len(x)
for i in range(int(val_count)):
  y_pred = c+ m*X[i]
  sst+= (y_pred-ybar)**2
  ssr+= (Y[i]-ybar)**2
rsquare = ssr/sst
print(1-rsquare)
```
